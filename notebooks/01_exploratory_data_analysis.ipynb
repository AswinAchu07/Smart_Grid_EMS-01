# 01 - Exploratory Data Analysis (EDA) for Smart Grid Microgrid

This notebook analyzes the physical correlations, diurnal cycles, seasonal patterns, and distributions of:
1. Solar Irradiance (GHI) and Solar PV Generation
2. Electricity Load Demand Profiles
3. Ambient Temperature, Cloud Cover, and Humidity
4. Time-of-Use (TOU) Pricing Structures

```python
import sys
sys.path.append('..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from app.data.synthetic_generator import generate_synthetic_dataset
from app.data.preprocessor import DataPreprocessor

df = generate_synthetic_dataset(periods=8760)
print('Dataset shape:', df.shape)
df.head()
```

```python
# Summary Statistics
df.describe()
```

```python
# 24-Hour Diurnal Profile Analysis
df['hour'] = pd.to_datetime(df['timestamp']).dt.hour
hourly_avg = df.groupby('hour')[['solar_power_kw', 'load_power_kw', 'temperature_c', 'irradiance_ghi_wm2']].mean()
hourly_avg.plot(figsize=(12, 6), grid=True, title='Average Diurnal Profiles')
plt.xlabel('Hour of Day')
plt.ylabel('Magnitude')
plt.show()
```
